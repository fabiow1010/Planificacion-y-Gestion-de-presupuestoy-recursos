## Cambio como elemento de Inovación
- Inovacion disrumtiva: Es porque en la empresa hacen una disrumcion de lo productos o servicios que ya estan y romper en el mercado en terminos de concepcion proyectando al publico con otro tipo de producto o elemento. Puede que el servicio no cambie en si, sino que cambie por completo lo que rodeo la prestacion del servicio.
    - Entre mejor se relacion y cooperacion de la empresa con el contexto/ entorno, es mas factible que innoven
      
- Inovacion evolutiva

       



Es de importante estar en la cadena del conocimiento, entraday salida del conocimiento para poder capturar esas ideas innovadoras que transforman procesos para ampliar los mnercads.

- Claves para la implementación de esas ideas inovadoras
inicialmente debemos capturar esas ideas, para ello debemos hacer un análisis de la informacion que me pueden dar los actores que se relacionan con el entorno de la compañia.
Claves que se pueden aplicar a los diferentes actores, tales como proveedores, clientes, competencia, mercados similares:
    * Aprovechar el mercado y toda la información que brinda
    * Estar atento al mundo de la investigacion a fin.
    * Adoptar estandar y regulacion
    * Explotar la propiedad intelectual
El fin de capturar la mayor de informacion es para materializar esas ideas inovadoras dado que cada idea trae detras todo un proyecto a implementar, vienen procesos de capacitacion, valdiacion, apopiación, porque esas ideas deben materializarse

Las ideas no se pueden quedar en solo eso, y la innovacion no es solo una buena idea, porque para que esa iniciativa llegue a materializarse debe tener una dimension del valor consistente, esta cadena del valor hay que saber "atacarla" de acuerdo a que actores beneficia cada proyecto/idea. 





Para identificar con claridad donde recae el valor debo cuantificarlo y cualificarlo. Medirlo en funcion de las variables que busca transformar el proyecto, puede ser en tiempo, costos... productividad etc.
hay que saber interpretar como percibe cada actor la cadena del valor, porque para cada actor se puede generar un valor diferente y hay que identificar como esta cadena del valor afecta los procesos, dependiendo del caracter del proceso puede lllegar a tener o no mayor prioridad dentro de la compañia.

## Enfoque hacia la mejora "Constante"





Metodologia **Lean** enfoque **DMAIC**
* **D**efinir
    * Definir la necesidad que tengo frente a un producto, diseños y procesos, cualificarla o cuantificar la problematica y que resultados se van a lograr cocn la implementacion de la idea. Definir la linea base del estado actual y la situacion deseada. Defino el alcance de mi iniciativa.
* **M**edir
    * Definir claramente como se realizara el proceso de medicion de varias: hay que definir
        * Método
        * Frecuencia
        * Umbrales

 

* **A**nalizar
    * Se trata de establecer puntos de control y recolectar datos con el fin de procesarlos para estimar desviacion o no conformidades sobre el proceso
    * Buscar establecer alertas tempranas sobre variables fuera de control que permitan tomar medidas de accion para solucionar los problemas que surgen en el marco del los procesos. 
* **M**ejorar
* **C**ontrolar
