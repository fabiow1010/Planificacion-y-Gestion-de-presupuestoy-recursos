## VPN(Valor Presente Neto) 
Sumatoria de los flujos de caja futuros menos el su valor presente utilizando una tasa de oportunidad.

**Un $VPN>0$ indica que el proyecto es economicamente viable y genera un valor neto positivo**
**VPN = 0 (Cero): El proyecto rinde exactamente lo mismo que la tasa de descuento; no crea ni destruye valor.**
**VPN < 0 (Negativo): El proyecto rinde menos de lo exigido; destruye valor y debe rechazarse**



* I₀: Es la inversión inicial realizada en el momento cero (t=0), expresada con un número negativo porque representa una salida de dinero.
* \(F_{t}\) Es el flujo de caja neto esperado en cada periodo t.
* r: Es la tasa de descuento (también llamada costo de capital o tasa mínima de rendimiento exigida), expresada en decimales.
* n: Es el número total de periodos que dura el proyecto (por lo general, años o meses).


## TIR (Tasa Interna de Retorno)
La fórmula de la TIR (Tasa Interna de Retorno) es aquella tasa de descuento (r) que hace que el Valor Actual Neto (VAN) de los flujos de caja de un proyecto sea igual a cero
Consiste en la tasa de descuento matematica que iguala el VPN a cero y representa la rentabildiad porcentual máxima.
**Se Considera viable si la TIR es mayor a la tasa de de descuento o costo de oportunidad exigido**



* I₀: Inversión inicial realizada (en negativo o restando al inicio).
* \(CF_{t}\)Flujo de caja (ingresos menos gastos) generado en cada periodo t.
* r: La tasa interna de retorno (la incógnita que se busca despejar).
* n o t: Número total de periodos (años, meses, etc.)

## Relacion Costro/Beneficio (Rerlacion B/C)
Consiste en dividir el valor presente de los ingresos proyectados entre el valor presente de los costos totales de inversion.



* B/C > 1: El proyecto es viable y rentable (los beneficios superan a los costos).
* B/C = 1: El proyecto está en punto de equilibrio (no genera ganancias ni pérdidas).
* B/C < 1: El proyecto no es rentable (los costos superan a los beneficios).
